```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0,20,500)
x = np.sin(t)

plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

##### POSITION VS TIME GRAPH
###### -- Periodic motion
###### -- Maximum displacement = amplitude

```python
v = -np.cos(t)
a = -np.sin(t)
plt.plot(x,v)
plt.xlabel("position")
plt.ylabel("velocity")
plt.show()
```
