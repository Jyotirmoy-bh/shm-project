```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,10,500)
A = 1   #amplitude
omega = 2   #angular frequency
phi = 0   #phase difference

x = A*np.cos(omega*t + phi)

plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("position")
plt.title("SHM Motion")
plt.show()
```

#### Standard SHM curve

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

x = np.zeros(len(t))
v = np.zeros(len(t))

# initial conditions
x[0] = 1
v[0] = 0

for i in range(len(t)-1):
    x[i+1] = x[i] + v[i]*dt
    v[i+1] = v[i] - x[i]*dt

plt.plot(t, x, label="Euler")
plt.legend()
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

#### Numerical method solution (Euler method)

```python
x_exact = np.cos(t)

plt.plot(t, x, label="Euler")
plt.plot(t, x_exact, label="Exact")
plt.legend()
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

x = np.zeros(len(t))
v = np.zeros(len(t))

# initial conditions
x[0] = 1
v[0] = 0

for i in range(len(t)-1):
    x[i+1] = x[i] + v[i]*dt
    v[i+1] = v[i] - x[i]*dt

plt.plot(t, v, label="Euler")
plt.legend()
plt.xlabel("time")
plt.ylabel("velocity")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

x = np.zeros(len(t))
v = np.zeros(len(t))

# initial conditions
x[0] = 1
v[0] = 0
b = 0.2   #damping parameter


for i in range(len(t)-1):
    a = -x[i] - b*v[i]
    x[i+1] = x[i] + v[i]*dt
    v[i+1] = v[i] + a*dt

plt.plot(t, x, label="damping")
plt.legend()
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

```python
plt.plot(x, v, label="damping")
plt.legend()
plt.ylabel("velocity")
plt.xlabel("position")
plt.show()
```

```python
E = 0.5*(x**2 + v**2)
plt.plot(t, E, label="damping")
plt.legend()
plt.xlabel("time")
plt.ylabel("energy")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

# comparing different effects of changing the value of b

b_values = [0.05, 0.2, 1]

for b in b_values:
    x = np.zeros(len(t))
    v = np.zeros(len(t))
    x[0] = 1
    v[0] = 0
    for i in range(len(t)-1):
        a = -x[i] - b*v[i]
        x[i+1] = x[i] + v[i]*dt
        v[i+1] = v[i] + a*dt
    plt.plot(t, x, label=f"b={b}")
plt.legend()
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

```python
for b in b_values:
    x = np.zeros(len(t))
    v = np.zeros(len(t))
    x[0] = 1
    v[0] = 0
    for i in range(len(t)-1):
        a = -x[i] - b*v[i]
        x[i+1] = x[i] + v[i]*dt
        v[i+1] = v[i] + a*dt
        E = 0.5*(x**2 + v**2)
    plt.plot(t, E, label=f"b={b}")
plt.legend()
plt.xlabel("time")
plt.ylabel("energy")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def simulate(dt, x_initial, v_initial):
    t = np.arange(0, 10, dt)

    x = np.zeros(len(t))
    v = np.zeros(len(t))

    x[0] = x_initial
    v[0] = v_initial

    for i in range(len(t)-1):
        x[i+1] = x[i] + v[i]*dt
        v[i+1] = v[i] - x[i]*dt

    return t, x, v


t, x, v = simulate(0.01, 1, 0)

plt.plot(t, x, label="Euler")
plt.legend()
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```
