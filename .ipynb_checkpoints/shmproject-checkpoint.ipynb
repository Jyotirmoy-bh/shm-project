```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0,20,500)
x = np.sin(t)

plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("position")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0,10,500)
A = 1
omega = 2
phi = 0

x = A*np.cos(omega*t + phi)

plt.plot(t,x)
plt.xlabel("time")
plt.ylabel("position")
plt.title("SHM Motion")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

x = np.zeros(len(t))
v = np.zeros(len(t))

# initial conditions
x[0] = 1
v[0] = 0

for i in range(len(t)-1):
    x[i+1] = x[i] + v[i]*dt
    v[i+1] = v[i] - x[i]*dt

plt.plot(t, x, label="Euler")
plt.legend()
plt.show()
```

```python
x_exact = np.cos(t)

plt.plot(t, x, label="Euler")
plt.plot(t, x_exact, label="Exact")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

dt = 0.01
t = np.arange(0, 10, dt)

x = np.zeros(len(t))
v = np.zeros(len(t))

# initial conditions
x[0] = 1
v[0] = 0
b = 0.2


for i in range(len(t)-1):
    a = -x[i] - b*v[i]
    x[i+1] = x[i] + v[i]*dt
    v[i+1] = v[i] + a*dt

plt.plot(t, x)
plt.show()
```

```python
plt.plot(x, v)
plt.show()
```

```python
E = 0.5*(x**2 + v**2)
plt.plot(t, E)
plt.show()
```
